## Applying Perturbation 

```python
from utils.optimizer import generate_pgd_dct_protection, generate_wscr_dct_protection
from utils.jpeg_dct import jpeg_decode, tensor_to_pil
from utils.data_controller import save_image_lossless
from utils.evaluation import advclean, jpeg_compress


iters = 350
eps = 1.0
alpha = 0.1

iters_10_percent = iters // 10
iters_50_percent = iters // 2
iters_70_percent = int(iters * 0.7)
checkpoints = [iters_10_percent, iters_50_percent, iters_70_percent]


img_protected_orig_paths = []
img_protected_purified_orig_paths = []
img_protected_compressed_orig_paths = []

iters_10_percent_protected_orig_paths = []
iters_50_percent_protected_orig_paths = []
iters_70_percent_protected_orig_paths = []

overall_runtimes = []
peak_gpu_memories = []

for (
    img_id,
    y,
    cb,
    cr,
    loss,
    overall_runtime,
    peak_gpu_memory,
    saved_checkpoints,
) in generate_wscr_dct_protection(
    iters,
    eps,
    alpha,
    current_batch,
    jpeg_quality,
    vae_encoder,
    frequency_weight=frequency_weight,
    checkpoints=checkpoints,
):
    img_src_orig_path = Path(f"{dataset_img_src_orig_path}/{img_id}_orig.png")
    img_src_edited_path = Path(f"{dataset_img_src_edited_path}/{img_id}_edited.png")

    img_protected_edited_path = Path(
        f"{results_protected_path}/orig/{img_id}_protected_orig.png"
    )
    img_protected_purified_edited_path = Path(
        f"{results_purified_path}/orig/{img_id}_protected_purified_orig.png"
    )
    img_protected_compressed_edited_path = Path(
        f"{results_compressed_path}/orig/{img_id}_protected_compressed_orig.png"
    )

    protected_img = tensor_to_pil(
        jpeg_decode(y, cb, cr, jpeg_quality=jpeg_quality, H=512, W=512, ste=False)
    )
    compressed_protected_img = jpeg_compress(
        protected_img, quality=jpeg_quality, verbose=True
    )
    purified_protected_img = advclean(compressed_protected_img)

    save_image_lossless(protected_img, img_protected_edited_path)
    save_image_lossless(compressed_protected_img, img_protected_purified_edited_path)
    save_image_lossless(purified_protected_img, img_protected_compressed_edited_path)

    img_protected_orig_paths.append(img_protected_edited_path)
    img_protected_compressed_orig_paths.append(img_protected_compressed_edited_path)
    img_protected_purified_orig_paths.append(img_protected_purified_edited_path)

    for chk_key in saved_checkpoints:
        chk_y, chk_cb, chk_cr = saved_checkpoints[chk_key].values()

        chk_save_path = Path(
            f"{results_protected_path}/chk/{chk_key}_{img_id}_protected_orig.png"
        )

        decoded = jpeg_decode(
            chk_y, chk_cb, chk_cr, jpeg_quality=jpeg_quality, H=512, W=512
        )
        save_image_lossless(
            tensor_to_pil(decoded),
            chk_save_path,
        )

        if chk_key == iters_10_percent:
            iters_10_percent_protected_orig_paths.append(chk_save_path)
        elif chk_key == iters_50_percent:
            iters_50_percent_protected_orig_paths.append(chk_save_path)
        elif chk_key == iters_70_percent:
            iters_70_percent_protected_orig_paths.append(chk_save_path)

    overall_runtimes.append(overall_runtime)
    peak_gpu_memories.append(peak_gpu_memory)

    log_message = f"{img_id} final loss at: {loss:.4f}\n\n"

    write_log(logger, log_message, verbose=True)

log_message = f"""[PERTURBATION PIPELINE]

- PGD iterations: {iters}
- Alpha: {alpha}
- Epsilon: {eps}

[INPUT BATCH]
- Number of images processed: {len(img_protected_orig_paths)}
- Image IDs:
{chr(10).join(f"  - {img_id}" for img_id in img_id_list) if img_id_list else "  (not populated yet)"}

"""

write_log(logger, log_message, verbose=True)
```

```python
import os
import psutil
from datetime import datetime
from IPython.display import display, HTML

# -------------------------


def alert_finished(start_timestamp):
    # Calculate Run Time (h m s)
    duration_seconds = time.time() - start_timestamp
    h, r = divmod(duration_seconds, 3600)
    m, s = divmod(r, 60)
    run_duration_str = f"{int(h):02d}h {int(m):02d}m {int(s):02d}s"

    # Calculate Server Uptime (h m s)
    p = psutil.Process(os.getppid())
    server_start_dt = datetime.fromtimestamp(p.create_time())
    uptime_delta = datetime.now() - server_start_dt

    total_uptime_seconds = uptime_delta.total_seconds()
    uh, ur = divmod(total_uptime_seconds, 3600)
    um, us = divmod(ur, 60)
    uptime_str = f"{int(uh):02d}h {int(um):02d}m {int(us):02d}s"

    formatted_server_start = server_start_dt.strftime("%I:%M %p")

    # JavaScript Template
    js_template = """
    <script>
    (function() {
        var doc = window.self !== window.top ? window.parent.document : document;
        var masterVolume = 0.7; 

        var backdrop = doc.createElement('div');
        backdrop.id = 'runFinishedBackdrop';
        backdrop.style.cssText = 'position:fixed;top:0;left:0;width:100vw;height:100vh;background:rgba(0,0,0,0.5);backdrop-filter:blur(3px);z-index:999999;display:flex;align-items:center;justify-content:center;cursor:pointer;';

        var container = doc.createElement('div');
        container.style.cssText = 'background:white;padding:40px;text-align:center;font-family:sans-serif;border-radius:12px;width:380px;box-shadow:0 10px 30px rgba(0,0,0,0.3);cursor:default;';

        container.innerHTML = `
            <div style="width:80px;height:80px;background:#28a745;border-radius:50%;display:flex;align-items:center;justify-content:center;margin:0 auto 20px auto;">
                <svg width="40" height="40" viewBox="0 0 24 24" fill="none" stroke="white" stroke-width="4" stroke-linecap="round" stroke-linejoin="round"><polyline points="20 6 9 17 4 12"></polyline></svg>
            </div>
            <h1 style="margin:0;color:#28a745;font-size:28px;">Run Finished</h1>
            <p style="color:#111;margin:15px 0 5px 0;font-size:20px;font-weight:bold;">Duration: {{DURATION}}</p>
            <p style="color:#888;font-size:13px;margin:0;">Server Up: {{UPTIME}} (Since {{START}})</p>
            <div style="margin-top:25px;">
                <button id="stopBtn" style="padding:12px 0;background:#28a745;color:white;border:none;border-radius:25px;cursor:pointer;font-size:16px;font-weight:bold;width:100%;">Dismiss</button>
            </div>
        `;

        backdrop.appendChild(container);
        doc.body.appendChild(backdrop);

        var context = new (window.AudioContext || window.webkitAudioContext)();
        var isLooping = true;
        var masterGain = context.createGain();
        masterGain.connect(context.destination);
        masterGain.gain.setValueAtTime(masterVolume, context.currentTime);

        function playChime() {
            if (!isLooping) return;
            [523.25, 659.25, 783.99].forEach(f => {
                var o = context.createOscillator();
                var g = context.createGain();
                o.type = 'sine';
                o.frequency.setValueAtTime(f, context.currentTime);
                g.gain.setValueAtTime(0, context.currentTime);
                g.gain.linearRampToValueAtTime(0.3, context.currentTime + 0.5);
                g.gain.exponentialRampToValueAtTime(0.001, context.currentTime + 3);
                o.connect(g).connect(masterGain);
                o.start(); o.stop(context.currentTime + 4);
            });
            setTimeout(playChime, 4500);
        }
        playChime();

        var close = function() {
            isLooping = false;
            context.close();
            backdrop.remove();
        };

        backdrop.onclick = close;
        doc.getElementById('stopBtn').onclick = close;
        container.onclick = e => e.stopPropagation();
    })();
    </script>
    """

    # Replace placeholders
    js_final = (
        js_template.replace("{{DURATION}}", run_duration_str)
        .replace("{{UPTIME}}", uptime_str)
        .replace("{{START}}", formatted_server_start)
    )

    display(HTML(js_final))


alert_finished(run_start_time)
```
