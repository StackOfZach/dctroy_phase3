```python
import requests
import os

url = "https://res.cloudinary.com/dajw2wgld/raw/upload/v1768984706/secrets_yhryeb.yaml"
# Define your specific path here.
# You can use absolute paths (C:/Users/...) or relative paths (./data/...)
save_path = f"configs/secrets.yaml"

# 1. Fetch the content
response = requests.get(url)
response.raise_for_status()  # Stop if the download failed

# 2. Ensure the directory exists (Optional but recommended)
# This prevents a "FileNotFoundError" if the folder doesn't exist yet
os.makedirs(os.path.dirname(save_path), exist_ok=True)

# 3. Write the content to the specific path
# We use 'wb' (write binary) to ensure exact encoding preservation
with open(save_path, "wb") as f:
    f.write(response.content)

print(f"File saved successfully at: {save_path}")

# 4. (Optional) Verify you can load it from the new local file
import yaml

with open(save_path, "r") as f:
    cfg = yaml.safe_load(f)
    print("Loaded YAML content:", cfg)
```
